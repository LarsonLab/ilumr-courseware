# What is MRI and How Does It Work?  A Science Fair Demonstration

The Ilumr tabletop MRI scanner is perfect for demonstrating MRI in a classroom or science fair setting.  It is small, portable, and safe to use, and can relatively quickly create images.  Running it through a laptop can allow participants to safely interact with scanning themselves.

## Ideas for a Science Fair

### Presentation Materials

A slideshow is included in this folder.  We also have created Printable Materials including posters and a MRI matching game handout here:

https://github.com/LarsonLab/MRI-education-resources/tree/main/Printable%20Materials

### Hands-on Scanning

Have a set of objects ready to scan.  These should be prepared ahead of time in vials, with examples of them placed on the table as well.

**Mystery mode**:  Put an object in scanner, and see if participants can identify

**Choice mode**: Let participants choose which one to scan, but guess what parts of the object they might be able to see before scanning.

**Interactive opportunities**
* Allow participants to click "Run" for scans
* Allow participants to scroll through a series of images
* Use Slice Viewer to view previous, high-resolution datasets
* Not recommended to have participants load samples as they may not appreciate the need to align the object correctly in the system

Recommend having simultaneous activities that participants can do.  For example, having matching game handout, a box with magnets and objects, and the scanner running at the same time.

## Objects for the Tabletop MRI Scanner

* Lego in water - see signal in water only.  Point out imaging of surface tension at top of the tube.

* Berries - look for seeds inside.  Blackberries are good since berries and seeds are larger.
* Onion - see layers, different image orientations
* Celery - good internal structure
* Broccoli - detailed structure
* flowers - detailed structure
* asparagus - detailed structure

* Orange - observe different layers in and around peel
* spider or small bugs?

## Sequences to run

* Calibrations
   * Find Frequency
   * (optional) Power Calibration
* Raw data and Signals (advanced)
   * FID - illustrate frequencies in signals, the signal decay
   * ``globals/frequency.yaml`` - set value and save to get a different frequency
* 2D Imaging with 2D RARE - do many Scans (averages) and will show how the image gets better.  Participants love to see if they can identify object as is accumulates
   * Parameter set - 64x64, 2-4mm slice (denser objects can be thinner slices), and 16-32 scans
* 3D Imaging - 3D RARE
   * Parameter set - 50x50x50, 4-8 scans, reasonable compromise of scan time and resolution, may require some Gaussian Blur
   * Scroll through slices (allow participants to do this too) to identify object
   * Change between ``F, P1, P2`` orientations to see different views of the object
* 3D scanned datasets with Slice Viewer - allow participants to scroll through previous datasets and identify objects
   * ``/data/default/`` has some good ones, like blackberry, cauliflower, lettuce
      * ``test1`` - blackberry, ``test2`` - cauliflower, ``test3`` - lettuce, ``test4`` - plant, ``test5`` - lego
