# What is MRI?

## How to run a Science Fair

Put mystery objects in scanner, and see if participants can identify

Allow participants to click "Run" for scans

Allow participants to scroll through a series of images

Use Slice Viewer to view previous, high-resolution datasets

## Objects for the Tabletop MRI Scanner

* Berries - look for seeds inside
* Onion - see layers, different image orientations
* Orange - some layers in peel
* Broccoli - nice structure
* flowers - nice structure
* asparagus - nice structure

* spider or small bugs?

## Sequences to run

* Calibration - Find Frequency
* Raw data - FID, illustrate frequencies in signals, the signal decay
* Initial and Fast 2D Imaging - 2D RARE, and do many Scans (averages) and will show how the image gets better.  Participants love to see if they can identify object as is accumulates
   * Parameter set - 64x64, 4mm slice (denser objects can be thinner slices), and 16 scans
* Initial and Fast 3D Imaging - 3D RARE, 50x50x50, 4 scans, reasonable compromise of scan time and resolution, may require some Gaussian Blur
* Higher resolution protocol?
