## Artifacts

* aliasing - change FOV.yaml
* ringing - turn off Gaussian filter 
* off-resonance displacement - 2D RARE in X-Y (circle object), change frequency.yaml for chemical shift, change shims.yaml (y shim for shear, x2z2 creates stretch/pileup!)
* susceptibility object artifacts?  3D FISP ok, was RARE better
* Ghosting - created these with ETL=NPE/2 with TR=1 in 2D RARE, then slide sample in and out between repetitions!
* chemical shift - this is very small, so far hard to see in spectrum, no image effect, maybe oil/water phantom would be better
* motion - ok with multi-shot RARE...
